```python
from pathlib import Path
import pandas as pd

# Path to PriceSense raw data
RAW_DIR = Path.home() / "Desktop" / "PriceSense" / "data" / "raw"

print("Raw data folder:")
print(RAW_DIR)
print()

# Check whether the folder exists
print("Folder exists:", RAW_DIR.exists())
print()

# List files
files = sorted(RAW_DIR.iterdir())

print("Files found:")
for file in files:
    print(f"{file.name:30} | {file.stat().st_size / (1024**2):,.2f} MB")

print("\n" + "=" * 70)

# Read only the first line of each file to inspect column headers
print("COLUMN HEADERS")
print("=" * 70)

for file in files:
    if file.is_file():
        with open(file, "r", encoding="utf-8-sig") as f:
            header = f.readline().strip()

        print(f"\n{file.name}")
        print(header)
```

```python
from collections import defaultdict
import numpy as np

# ---------------------------------------------------------
# Day 1 — Structural profiling
# ---------------------------------------------------------

profile = {}

for file in sorted(RAW_DIR.glob("*.csv")):

    print(f"\nProfiling: {file.name}")

    # Read in chunks so large files do not overwhelm RAM
    chunksize = 500_000

    row_count = 0
    null_counts = defaultdict(int)

    min_date = None
    max_date = None

    unique_items = set()
    unique_stores = set()

    for chunk in pd.read_csv(
        file,
        chunksize=chunksize,
        low_memory=False
    ):

        row_count += len(chunk)

        # Remove exported index column from profiling
        if "Unnamed: 0" in chunk.columns:
            chunk = chunk.drop(columns="Unnamed: 0")

        # Null counts
        for col in chunk.columns:
            null_counts[col] += int(chunk[col].isna().sum())

        # Date range
        if "date" in chunk.columns:
            dates = pd.to_datetime(chunk["date"], errors="coerce")

            chunk_min = dates.min()
            chunk_max = dates.max()

            if pd.notna(chunk_min):
                min_date = chunk_min if min_date is None else min(min_date, chunk_min)

            if pd.notna(chunk_max):
                max_date = chunk_max if max_date is None else max(max_date, chunk_max)

        # Unique items
        if "item_id" in chunk.columns:
            unique_items.update(chunk["item_id"].dropna().unique())

        # Unique stores
        if "store_id" in chunk.columns:
            unique_stores.update(chunk["store_id"].dropna().unique())

    profile[file.name] = {
        "rows": row_count,
        "columns": list(null_counts.keys()),
        "null_counts": dict(null_counts),
        "date_min": min_date,
        "date_max": max_date,
        "unique_items": len(unique_items),
        "unique_stores": len(unique_stores)
    }

# ---------------------------------------------------------
# Display summary
# ---------------------------------------------------------

summary = []

for filename, info in profile.items():
    summary.append({
        "file": filename,
        "rows": info["rows"],
        "columns": len(info["columns"]),
        "date_min": info["date_min"],
        "date_max": info["date_max"],
        "unique_items": info["unique_items"],
        "unique_stores": info["unique_stores"]
    })

summary_df = pd.DataFrame(summary)

display(summary_df)

print("\n" + "=" * 70)
print("NULL VALUES")
print("=" * 70)

for filename, info in profile.items():
    print(f"\n{filename}")

    null_df = pd.DataFrame(
        list(info["null_counts"].items()),
        columns=["column", "null_count"]
    )

    null_df = null_df[null_df["null_count"] > 0]

    if null_df.empty:
        print("No null values.")
    else:
        display(null_df)
```
